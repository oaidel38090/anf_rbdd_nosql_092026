# Plan pédagogique de la formation “Vous avez dit NoSQL”

## Présentation de la MITI et de rBDD (30min)

## 1 – Contexte et évolution des bases de données (Osman AIDEL ) (1h30)

:::{important}:Objectif
Comprendre l’évolution des systèmes de gestion de bases de données et les limites du modèle relationnel dans certains contextes.
:::

* Historique et évolution des bases de données
* Rappels sur les bases relationnelles : atouts, limites, usages
* Nouveaux défis liés à la volumétrie, la variabilité et la vitesse des données
* Émergence des bases NoSQL et tendances actuelles

## 2 – Comprendre l’approche NoSQL (Juba AGOUN) (1h30)

:::{important}:Objectif
Poser les bases théoriques du NoSQL et situer les différentes familles de bases.
:::

* La mouvance NoSQL
* Le théorème de CAP et ses implications
* Modélisation des données dans un paradigme NoSQL
* Présentation des principales familles : document, graphe, clé-valeur, colonne
* Étude de cas et comparaison avec le SQL : pour chaque modèle, mise en lumière des avantages/inconvénients face au relationnel

## 3 – Mise en place de l’environnement de travail (Osman AIDEL) (1h00)

:::{important}:Objectif
Prendre en main l’environnement des TP.
:::

* Introduction à Docker et ses usages en formation
* Lancement et configuration des conteneurs pour chaque technologie
* Prise en main des interfaces et outils de requêtage

## 4 – Bases de données orientées graphe (Neo4j) (Juba AGOUN) (3h00)

:::{important}:Objectif
Découvrir la modélisation en graphe et apprendre les bases du langage Cypher.
:::

* Présentation des concepts de graphe (nœuds, relations, propriétés)
* Comparaison avec les tables relationnelles
* Cas d’usage typiques
* Introduction aux opérations CRUD (Cypher) en Neo4j
* Atelier pratique : modélisation et requêtage avec Neo4j (Jeu de données : Game of Thrones)

## 5 – Bases de données orientées document (MongoDB) (Osman AIDEL) (3h00)

:::{important}:Objectif
Comprendre le modèle document, le langage de requête MongoDB et ses cas d’usage.
:::

* Présentation du modèle document
* Comparaison avec les tables relationnelles
* Cas d’usage typiques
* Introduction aux opérations CRUD en MongoDB
* Atelier pratique : manipulation de données avec MongoDB (illustrations sur jeux de données dédiés)

## 6 – Bases de données clé-valeur (Redis) (Osman AIDEL) (3h00)

:::{important}:Objectif
Explorer les usages et performances des bases clé-valeur dans des architectures à forte contrainte.
:::

* Présentation du modèle clé-valeur
* Comparaison avec les tables relationnelles
* Cas d’usage typiques
* Introduction aux opérations CRUD en Redis
* Atelier pratique : manipulation de clés et structures avancées dans Redis

## 7 – Bases de données orientées colonnes (Cassandra / ClickHouse) (Juba AGOUN) (3h00)

:::{important}:Objectif
Comprendre les modèles en colonnes et les cas d’usage adaptés aux très grands volumes.
:::

* Présentation du modèle en colonnes
* Comparaison avec les tables relationnelles
* Cas d’usage typiques
* Introduction aux opérations CRUD (CQL) en Cassandra et focus ClickHouse
* Atelier pratique : manipulation de données orientées BI avec Cassandra / ClickHouse

## 8 – Architectures distribuées : réplication (Osman AIDEL) (2h30)

:::{important}:Objectif
Comprendre les mécanismes de réplication pour la disponibilité et les performances.
:::

* Notions de réplication, de quorum et de cohérence (consistency)
* Exemples de configuration de réplication avec MongoDB
* La réplication au profit de la performance
* Réflexion sur l'architecture Peer-to-Peer
* Atelier pratique : mise en place d’un cluster répliqué

## 9 – Architectures distribuées : sharding (partitionnement horizontal) (Osman AIDEL) (3h00)

:::{important}:Objectif
Comprendre et expérimenter la scalabilité horizontale dans les systèmes NoSQL.
:::

* Principe du sharding et de la distribution des données
* Atelier pratique : configuration d’un cluster sharded
