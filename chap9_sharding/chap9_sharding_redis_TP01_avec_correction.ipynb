# TP REDIS Clustering



## Introduction — Sharding avec Redis Cluster

Lorsque le volume de données ou la charge devient trop important pour être traité efficacement par une seule instance Redis, il devient nécessaire de répartir les données entre plusieurs nœuds. Cette technique de partitionnement horizontal est appelée sharding.

Redis met en œuvre ce mécanisme à travers Redis Cluster. L’espace des clés est divisé en 16 384 hash slots, répartis entre les différents nœuds primaires du cluster. Pour chaque clé, Redis détermine son hash slot afin d’identifier le nœud responsable de son stockage.

Dans ce TP, nous allons construire un cluster composé de 3 nœuds primaires et 3 replicas. Nous observerons concrètement comment Redis distribue les clés, comment les requêtes sont dirigées vers le bon nœud et comment le cluster réagit à la perte d’un nœud primaire.

:::{important} Objectifs du TP

À l’issue du TP, vous serez capables de :

* déployer et initialiser un Redis Cluster ;
* identifier les rôles Primary / Replica ;
* comprendre la répartition des 16 384 hash slots ;
* déterminer le slot associé à une clé ;
* observer la distribution des données entre les nœuds ;
* comprendre le mécanisme de failover lors de la perte d’un Primary ;
* utiliser les Hash Tags pour placer des clés dans un même slot ;
* comprendre l’origine d’une erreur CROSSSLOT.

:::

## Hash Tags : conserver plusieurs clés dans le même slot

Redis Cluster permet de créer une **affinité entre plusieurs clés** afin qu'elles soient placées dans le même hash slot. Cette fonctionnalité est particulièrement utile lorsque plusieurs clés représentent des données liées et doivent être utilisées ensemble.

Commençons par observer deux clés classiques :

```redis
CLUSTER KEYSLOT user:1:name
CLUSTER KEYSLOT user:1:age
```

Ces deux clés peuvent être associées à des slots différents.

Pour forcer Redis à utiliser la même partie de la clé lors du calcul du slot, on utilise un **Hash Tag**, c'est-à-dire une chaîne placée entre accolades `{}` :

```redis
CLUSTER KEYSLOT user:{1}:name
CLUSTER KEYSLOT user:{1}:age
```

Dans les deux cas, Redis utilise uniquement `1` pour calculer le hash slot :

```text
user:{1}:name ──┐
                ├── hash("1") ──► même hash slot
user:{1}:age  ──┘
```

> **À retenir :** dans une clé contenant `{...}`, Redis Cluster utilise le contenu entre accolades pour déterminer le hash slot.


#### Créez maintenant 2 nouvelles clés `user:1:name` et `user:1:age` avec les valeurs respective Alice et 32. Les 2 clés devront être sotckées dans le même shard ?


:::{admonition} Solution
:class: dropdown
```
    set user:1:name{user:1} Alice
    set user:1:name{user:1} 32
```

#### Retourner la valeur de vos nouvelles clé ?

:::{admonition} Solution
:class: dropdown

`GET user:1:name` ne fonctionne pas car `user:1:name` et `user:1:name{user:1}` sont deux noms de clés différents.

Il faut utiliser le nom complet de la clé :

```redis
GET user:1:name{user:1}
```

Le Hash Tag influence le **calcul du slot**, mais les accolades et leur contenu font toujours partie du nom de la clé.
:::

## Opérations multi-clés et erreur CROSSSLOT

Les Hash Tags deviennent particulièrement importants lorsqu'une commande manipule **plusieurs clés**. Dans Redis Cluster, certaines opérations multi-clés nécessitent que toutes les clés concernées appartiennent au **même hash slot**.

Testez d'abord :

```redis
SET user:1:name Alice
SET user:1:city Lyon
CLUSTER KEYSLOT user:1:name
CLUSTER KEYSLOT user:1:city
MGET user:1:name user:1:city
```

Si les deux clés appartiennent à des slots différents, Redis retourne une erreur de type :

```text
CROSSSLOT Keys in request don't hash to the same slot
```

Elle apparaît lorsqu’une commande veut manipuler plusieurs clés qui ne se trouvent pas dans le même hash slot.

Utilisez maintenant un Hash Tag :

```redis
SET user:{1}:name Alice
SET user:{1}:city Lyon
CLUSTER KEYSLOT user:{1}:name
CLUSTER KEYSLOT user:{1}:city
MGET user:{1}:name user:{1}:city
```

#### Que constatez-vous ? Pourquoi l'utilisation de `{1}` permet-elle maintenant d'exécuter la commande `MGET` ?

> Les deux clés utilisent le même Hash Tag `{1}` : elles appartiennent donc au même hash slot et peuvent être traitées ensemble par le même shard.

:::{admonition} Solution
:class: dropdown

Le nœud primaire précédemment arrêté est maintenant considéré comme réparé. Vous pouvez redémarrer le conteneur `redis-node-3` afin qu'il réintègre le cluster.
:::

#### Que constatez vous ?

:::{admonition} Solution
:class: dropdown

Le nœud `172.21.0.9` a réintégré le cluster en tant que **replica** du nouveau primaire `172.21.0.6`.

Le failover n'est donc pas automatiquement annulé au retour de l'ancien primaire : le nœud qui a été promu reste primaire et l'ancien primaire rejoint ici le cluster comme replica.
:::

# Conclusion

Dans ce TP, nous avons mis en œuvre un Redis Cluster afin de comprendre concrètement le fonctionnement du sharding. Les données ne sont plus stockées sur une seule instance.

Nous avons observé le chemin suivi par une donnée :

Clé → Hash Slot → Nœud primaire → Replica

Cette répartition permet à Redis de distribuer les données et la charge entre plusieurs nœuds. La réplication complète ce mécanisme en apportant de la haute disponibilité : lorsqu’un primaire devient indisponible, un replica peut être promu et prendre son rôle.

Nous avons également constaté que le sharding introduit de nouvelles contraintes. Une commande s’exécute sur le nœud responsable du slot concerné et certaines opérations portant sur plusieurs clés nécessitent que celles-ci soient placées dans le même hash slot. Les Hash Tags permettent de contrôler cette colocalisation et d’éviter notamment les erreurs CROSSSLOT.

À retenir : Redis Cluster combine sharding pour répartir les données et la charge, et réplication pour assurer la disponibilité. Le choix du nom des clés devient alors important, car il peut directement influencer leur répartition et les opérations qu’il sera possible d’effectuer ensemble.

Cette conclusion est particulièrement cohérente avec ton TP parce qu’elle reprend exactement les quatre choses que les étudiants ont manipulées : slots → distribution → failover → Hash Tags/CROSSSLOT.
