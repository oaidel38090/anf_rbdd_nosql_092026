# TP Neo4j : Prise en main, manipulation et requêtage



L'objectif de ce TP est de mettre en pratique le requêtage via une algèbre en utilisant **Cypher** dans une base de données Neo4j.

Dans une **première partie**, nous installerons Neo4j via Docker et vérifierons que l'installation fonctionne.

Dans une **seconde partie**, nous utiliserons une base de données de films et effectuerons différentes requêtes Cypher.

Dans une **troisième partie**, nous découvrirons la modélisation dans Neo4j avec l'univers de *Game of Thrones*.

# Annexe : mémo Docker utile

| Besoin | Commande |
|---|---|
| Voir si le conteneur tourne | `docker ps` |
| Voir les logs | `docker logs neo4j-tp` |
| Arrêter le conteneur | `docker stop neo4j-tp` |
| Relancer le conteneur | `docker start neo4j-tp` |
| Arrêter et supprimer (Compose) | `docker compose down` |
| Tout réinitialiser (Compose) | `docker compose down -v` |
| Ouvrir un shell Cypher en ligne de commande | `docker exec -it neo4j-tp cypher-shell -u neo4j -p anfneo4j2026` |
| Le port 7474/7687 est déjà utilisé | Changez la partie gauche du mapping, par exemple `-p 7475:7474 -p 7688:7687`, puis adaptez l'URL de connexion. |
