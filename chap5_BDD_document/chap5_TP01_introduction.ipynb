# Introduction à MONGODB

L'objectif de ce notebook est de vous familiariser avec votre environnement MongoDB. 


# 1. Préparer l'environnement

Depuis la racine du dépôt, placez-vous dans `docker`. Démarrez le service et vérifiez son état :

:::bash
cd docker
docker compose up -d mongosingle
docker compose ps mongosingle
:::

Vous venez de démarrer un serveur mongo qui tourne dans un container

Pour vérifier que votre container mongo est opérationnel, vous pouvez utiliser la commande docker :

:::bash
docker compose ps mongosingle
:::
Résultat obtenu :
```
CONTAINER ID   IMAGE     COMMAND                  CREATED         STATUS         PORTS                                           NAMES
bca94fa31708   mongo     "docker-entrypoint.s…"   7 seconds ago   Up 5 seconds   0.0.0.0:27017->27017/tcp, :::27017->27017/tcp   mongosingle
```
