# TP MongoDB — Passez de la recherche à l’analyse l'Aggregation Framework!

Dans le TP précédent, vous avez appris à rechercher, filtrer, trier et modifier des documents avec MQL.

Vous savez maintenant répondre à des questions comme :
* Quels sont les films réalisés par Clint Eastwood ?
* Quels films ont été produits après une certaine année ?

Mais que se passe-t-il lorsque nous voulons analyser un ensemble de documents ?

* Combien de documents avons-nous par catégorie ?
* Quelle est la somme d’une valeur pour chaque groupe ?
* Quelle est la moyenne, le minimum ou le maximum ?
* Comment obtenir un Top 10 ?
* Et surtout… comment réaliser l’équivalent d’un GROUP BY SQL avec MongoDB ?

C’est précisément le rôle de l’Aggregation Framework.

MongoDB permet de construire un pipeline d’agrégation : les documents traversent successivement différentes étapes, appelées stages, chacune réalisant une opération.

```
Documents
    │
    ▼
 $match       🔎 Filtrer
    │
    ▼
 $group       📦 Regrouper et calculer
    │
    ▼
 $sort        ↕️ Trier
    │
    ▼
 $limit       ✂️ Limiter
    │
    ▼
 Résultat
```

:::{important} Vos objectifs techniques

À la fin de ce TP, vous serez capables de :

* Construire un pipeline d’agrégation avec aggregate()
* Filtrer les documents avec $match
* Regrouper des documents selon un ou plusieurs critères avec $group
* Réaliser des calculs avec $sum, $avg, $min et $max
* Trier les résultats avec $sort
* Limiter les résultats avec $limit
* Enchaîner plusieurs stages pour construire progressivement une analyse
* Comprendre pourquoi l’ordre des stages est important

::: 

## Le principe à retenir

Un pipeline se construit étape par étape.



Chaque stage reçoit les documents produits par le stage précédent, les transforme, puis transmet le résultat au suivant.

À vous maintenant de faire parler les données !

La liste des étapes est disponible à l'url suivante :

https://docs.mongodb.com/manual/reference/operator/aggregation-pipeline/

## Introduction

Chargez les données suivantes dans la base de données training dans la collection cats : 

```
db.cats.insertMany([
{ _id: 1, name: "Fluffy", type: "Cat", sex : "M", weight: 5 },
{ _id: 2, name: "Scratch", type: "Cat", sex : "M", weight: 3 },
{ _id: 3, name: "Spike", type: "Cat", sex : "M", weight: 6 },
{ _id: 4, name: "Amber", type: "Cat", sex : "F", weight: 7 },
{ _id: 5, name: "Lady", type: "Cat", sex : "F", weight: 4 },
{ _id: 6, name: "Smokey", type: "Cat", sex : "F", weight: 4 } 
])
```


Nous allons à présent utiliser le framework, pour calculer le nombre d'animaux par genre lancez la commande suivante dans NoSQLBooster :

:::{note}Note
L'opération $group regroupe les données par la clé _id et pour tous les documents dont _id est identique, le framework applique la fonction donnée en 2nd argument. 
:::

Le résultat retourne les valeurs suivantes :

Essayons à présent de trier les données obtenues en sortie du $group par genre :

Filtrez le résultat précédent en ne sélectionnant que les animaux dont le poids est > 3.

:::{admonition} Solution
:class: dropdown
```javascript
db.cats.aggregate([
  { $match: { weight: { $gt: 3 } } },
  { $group: { _id: "$sex", count: { $sum: 1 } } },
  { $sort: { _id: 1 } }
])
```
:::

Ajoutez sur la requête précédente le calcul de la moyenne des poids :

:::{admonition} Solution
:class: dropdown
```javascript
db.cats.aggregate([
  { $match: { weight: { $gt: 3 } } },
  {
    $group: {
      _id: "$sex",
      count: { $sum: 1 },
      avg: { $avg: "$weight" }
    }
  },
  { $sort: { _id: 1 } }
])
```
:::

## Exercice

Pour cet exercice, vous travaillerez sur la collection zips.
Cette collection contient les codes postaux des états unis sous la structure suivante :

#### A partir du pipeline d'agrégation, calculer la population de chaque état :

:::{admonition} Solution
:class: dropdown
```javascript
db.zips.aggregate([
  { $group: { _id: "$state", population: { $sum: "$pop" } } }
])
```
:::

#### Triez le résultat de la question précédente par ordre décroissant du nombre d'habitants.

:::{admonition} Solution
:class: dropdown
```javascript
db.zips.aggregate([
  { $group: { _id: "$state", population: { $sum: "$pop" } } },
  { $sort: { population: -1 } }
])
```
:::

#### Trouvez les 10 plus grandes villes CAliforniennes : 

:::{admonition} Solution
:class: dropdown
```javascript
db.zips.aggregate([
  { $match: { state: "CA" } },
  { $group: { _id: "$city", population: { $sum: "$pop" } } },
  { $sort: { population: -1 } },
  { $limit: 10 }
])
```
:::

#### Identifiez les états dont la population est supérieure à 10 millions et les trier par ordre décroissant de population  : 

:::{admonition} Solution
:class: dropdown
```javascript
db.zips.aggregate([
  { $group: { _id: "$state", population: { $sum: "$pop" } } },
  { $match: { population: { $gt: 10000000 } } },
  { $sort: { population: -1 } }
])
```
:::

#### Calculez la population que représente les villes Californiennes de plus de 500 000 habitants :

:::{admonition} Solution
:class: dropdown
```javascript
db.zips.aggregate([
  { $match: { state: "CA" } },
  { $group: { _id: "$city", population: { $sum: "$pop" } } },
  { $match: { population: { $gt: 500000 } } },
  { $group: { _id: null, population: { $sum: "$population" } } }
])
```
:::

#### Calculez pour chaque état et par région (2 premiers digits du zip code)  la population, le nombre moyen d'habitants, le nombre maximum et minimum d'habitants  : 
* utilisez l'operateur ```substr``` pour le traitement du zip code.

https://docs.mongodb.com/manual/reference/operator/aggregation/substr/#:~:text=%24substr%20(aggregation)%C2%B6&text=Returns%20a%20substring%20of%20a,The%20index%20is%20zero%2Dbased.&text=The%20arguments%20can%20be%20any,third%20arguments%20resolve%20to%20integers.



:::{admonition} Solution
:class: dropdown
```javascript
db.zips.aggregate([
  { $match: { pop: { $gt: 50000 } } },
  {
    $group: {
      _id: {
        state: "$state",
        region: { $substr: ["$zip", 0, 2] }
      },
      sumPop: { $sum: "$pop" },
      avgPop: { $avg: "$pop" },
      minPop: { $min: "$pop" },
      maxPop: { $max: "$pop" }
    }
  }
])
```
:::
